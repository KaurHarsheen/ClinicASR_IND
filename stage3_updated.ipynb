```python
from kokoro import KPipeline
import soundfile as sf
import os
import json
import numpy as np
# 3. Initialize Pipeline (Forced to CPU)
pipeline = KPipeline(lang_code='a', device='cpu') 
voice = 'af_heart' 
prepared_data = []



# 4. Process scripts
for i, script in enumerate(medical_scripts):
    # Path points to your local Jupyter directory now
    file_name = f"sample_{i}.wav"
    output_path = os.path.join(save_directory, file_name)
    
    # Generate and concatenate audio
    generator = pipeline(script, voice=voice, speed=1.0)
    audio_segments = [audio for _, _, audio in generator]
    
    # Safety check to ensure audio was actually generated
    if audio_segments: 
        combined_audio = np.concatenate(audio_segments)
        
        # Save to file
        sf.write(output_path, combined_audio, 24000)
        
        # Append to list using Hugging Face's required keys
        prepared_data.append({"file_name": file_name, "transcription": script})
        print(f"Synthesized: {output_path}")

# 5. Save the metadata as JSONL (Standard for Hugging Face AudioFolder)
metadata_path = os.path.join(save_directory, "metadata.jsonl")
with open(metadata_path, 'w', encoding='utf-8') as f:
    for item in prepared_data:
        f.write(json.dumps(item) + '\n')

print("-" * 30)
print(f"Batch generation complete!")
print(f"Audio files and metadata.jsonl are saved in: {save_directory}")
```

```python
import os

# ---------------------------------------------------------
# 0. HARDWARE ISOLATION & MEMORY PATCHES
# ---------------------------------------------------------
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import torch
torch.backends.cudnn.benchmark = False
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.enabled = False

from dataclasses import dataclass
from typing import Any, Dict, List, Union
from transformers import (
    WhisperForConditionalGeneration, 
    WhisperProcessor, 
    Seq2SeqTrainingArguments, 
    Seq2SeqTrainer,
    BitsAndBytesConfig
)
from peft import get_peft_model, LoraConfig
from datasets import load_dataset, Audio

# ---------------------------------------------------------
# 1. SETUP PATHS & LOAD BASE MODEL + FRESH LORA ADAPTER
# ---------------------------------------------------------
model_id = "openai/whisper-small"

processor = WhisperProcessor.from_pretrained(model_id, language="Hindi", task="transcribe")
quant_config = BitsAndBytesConfig(load_in_8bit=True)
from peft import PeftModel
model = WhisperForConditionalGeneration.from_pretrained(
    model_id,
    quantization_config=quant_config,
    device_map="auto"
)
model.config.forced_decoder_ids = None
model.config.suppress_tokens = []
model.enable_input_require_grads()
model.config.use_cache = False

checkpoint_dir = "./medical_whisper_final" # Ya "./my_whisper_checkpoint/checkpoint-200"

# is_trainable=True ensures ki model freeze na ho aur nayi training ke liye ready rahe
model = PeftModel.from_pretrained(model, checkpoint_dir, is_trainable=True)

print("Imports, GPU 0 Isolation, and Fresh Model Loading Complete!")
```

```python
import json
import os
import librosa
from datasets import Dataset
from tqdm.auto import tqdm

# ---------------------------------------------------------
# 2. LOAD & PREPROCESS (NATIVE DISK LOADING)
# ---------------------------------------------------------
print("Loading metadata from disk...")

data_dir = "./Medical_Audio_Dataset"
metadata_path = os.path.join(data_dir, "metadata.jsonl")
processed_data = []

# 1. Read the JSONL file directly
with open(metadata_path, 'r', encoding='utf-8') as f:
    for line in tqdm(f, desc="Processing Audio Files"):
        item = json.loads(line)
        
        # 2. Construct the absolute path to the .wav file
        audio_path = os.path.join(data_dir, item["file_name"])
        
        # 3. Load the audio using librosa (forces 16kHz automatically)
        # Librosa does not rely on torchcodec or ffmpeg!
        audio_array, _ = librosa.load(audio_path, sr=16000)
        
        # Ensure it's mono
        if len(audio_array.shape) > 1:
            audio_array = librosa.to_mono(audio_array.T)
            
        # 4. Extract features directly via WhisperProcessor
        input_features = processor.feature_extractor(
            audio_array, sampling_rate=16000
        ).input_features[0]
        
        # Map the transcription to tokens
        labels = processor.tokenizer(item["transcription"]).input_ids 
        
        # Store in our list
        processed_data.append({
            "input_features": input_features,
            "labels": labels
        })

print("Converting to Hugging Face Dataset format...")
# Convert the natively processed data back into a HF Dataset for the trainer
medical_dataset = Dataset.from_list(processed_data)

print("Dataset ready for training!")
```

```python
# ---------------------------------------------------------
# 3. DATA COLLATOR
# ---------------------------------------------------------
@dataclass
class DataCollatorSpeechSeq2SeqWithPadding:
    processor: Any

    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        # Pad input features (audio)
        input_features = [{"input_features": feature["input_features"]} for feature in features]
        batch = self.processor.feature_extractor.pad(input_features, return_tensors="pt")

        # Pad labels (text tokens)
        label_features = [{"input_ids": feature["labels"]} for feature in features]
        labels_batch = self.processor.tokenizer.pad(label_features, return_tensors="pt")

        # Replace padding with -100 so the loss function ignores it
        labels = labels_batch["input_ids"].masked_fill(labels_batch.attention_mask.ne(1), -100)
        
        # Strip the BOS token if present
        if (labels[:, 0] == self.processor.tokenizer.bos_token_id).all().cpu().item():
            labels = labels[:, 1:]

        batch["labels"] = labels
        return batch

data_collator = DataCollatorSpeechSeq2SeqWithPadding(processor=processor)
print("Data Collator initialized!")
```

```python
# ---------------------------------------------------------
# 4. LOCAL GPU OPTIMIZED TRAINING ARGUMENTS
# ---------------------------------------------------------
training_args = Seq2SeqTrainingArguments(
    output_dir="./medical_whisper_model",
    per_device_train_batch_size=8,  
    gradient_accumulation_steps=4,  
    learning_rate=1e-4,             
    warmup_steps=50,
    max_steps=400,                  
    bf16=False,                     # <-- DISABLED for local GPU compatibility
    fp16=True,                      # <-- ENABLED for local GPU compatibility
    optim="adamw_8bit",
    eval_strategy="no",             
    save_steps=100,
    logging_steps=10,
    report_to=["none"],
    remove_unused_columns=False,
    label_names=["labels"],
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={"use_reentrant": True},
)

trainer = Seq2SeqTrainer(
    args=training_args,
    model=model,
    train_dataset=medical_dataset,
    data_collator=data_collator,
    processing_class=processor, 
)
print("Training arguments successfully configured for local GPU!")
```

```python
# ---------------------------------------------------------
# 5. START TRAINING & SAVE
# ---------------------------------------------------------
print("Starting fine-tuning...")
trainer.train()

print("Saving final medical model and processor...")
# Save the LoRA weights
trainer.save_model("./medical_whisper_final")

# CRITICAL: Save the processor so you can easily load this later in production
processor.save_pretrained("./medical_whisper_final")

print("✅ Training complete! Model is ready for inference.")
```
